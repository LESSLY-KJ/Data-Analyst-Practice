```python
def calculate_salary(basic_salary):
    bonus = basic_salary*10/100
    deduction=basic_salary*5/100
    final_salary=basic_salary+bonus-deduction
    return final_salary
salary=float(input("Enter Salary:"))
result=calculate_salary(salary)
print("Final Salary",result)
```

```python
students = {
    "Arun": 85,
    "Bala": 45,
    "Kumar": 90,
    "Ravi": 35,
    "Sam": 75
} 
print("All students:")
for name,marks in students.items():
 print(name,":",marks)

print("\nPassed Students:")
for name,marks in students.items():
     if (marks>=50): 
      print(name,":",marks) 
print("\nFailed Student:")
for name,marks in students.items():
  if(marks<50):
    print(name,":",marks)

total=sum(students.values())      
avg=total/len(students)
print("\nAvarege Marks:",avg) 

topper=max(students,key=students.get)
print("\nTopper:",topper)
print("TopperMarks:",students[topper])

```
