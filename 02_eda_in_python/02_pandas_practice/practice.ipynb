```python
def calculate_salary(basic_salary):
    bonus = basic_salary*10/100
    deduction=basic_salary*5/100
    final_salary=basic_salary+bonus-deduction 
    return final_salary
salary=float(input("Enter Salary:"))
result=calculate_salary(salary)
print("Final Salary",result)
```
