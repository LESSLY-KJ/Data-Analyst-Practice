### 📥 Pandas Items 01-05: Dataset Loading & Initial Preview
- **Dataset:** `sales_data.csv` (E-Commerce orders)
- **Concepts:** `pd.read_csv()`, `df.head()`, `df.tail()`, `df.info()`

```python
import pandas as pd

# Create a sample CSV file with a subtle missing value & data issue
csv_data = """order_id,customer_name,category,amount,city
101,Alice,Electronics,1200.50,New York
102,Bob,Clothing,80.00,Chicago
103,Charlie,Electronics,450.00,New York
104,David,Groceries,,Chicago
105,Emma,Clothing,200.00,Miami
106,Frank,Electronics,950.00,San Francisco
107,Grace,Groceries,120.00,Boston
"""

with open("sales_data.csv", "w") as f:
    f.write(csv_data)

print("sales_data.csv created successfully!")

df=pd.read_csv("sales_data.csv") 

print("-------First Three Row-------")
display(df.head(3))
print("-------Last Two Row-------")
display(df.tail(2))
print("\n--- DATASET INFO ---")
display(df.info())  
```

### 🔗 Pandas Items 19-20: Table Merging & Joining (SQL JOIN in Python)
- **Reference Tables:** `discounts` (Category discounts) & `shipping_status`
- **Concepts:** `pd.merge()` on column (`inner`, `left`, `outer`, `right`), `df.join()` on index

```python
# A small reference table for discounts
discounts = pd.DataFrame({
    "category": ["Electronics", "Clothing", "Groceries"],
    "discount_pct": [10, 20, 5]
})
discounts
```

```python
df.groupby("category")["amount"].agg(["min","max","mean","median","sum"])
```

```python
df.sort_values(by="amount",ascending=False)
```

```python
df["location"].value_counts()
```

```python
pd.merge(df,discounts,on="category",how="inner")
```

```python
pd.merge(df,discounts,on="category",how="left")
```

```python
pd.merge(df,discounts,on="category",how="outer")
```

```python
pd.merge(df,discounts,on="category",how="right")
```

```python
# 1. Create a table with 7 shipping statuses
shipping_status = pd.DataFrame({
    "status": ["Delivered", "Shipped", "Delivered", "Pending", "Shipped", "Delivered", "Pending"]
})

# 2. Join it directly to df!
df.join(shipping_status)
```
