```python
import sqlite3
import pandas as pd

# Setup a local practice database
conn = sqlite3.connect(':memory:')

# Create a sample World table to match SQLZoo
conn.execute('''
CREATE TABLE world (
    name TEXT,
    continent TEXT,
    area INTEGER,
    population INTEGER,
    gdp INTEGER
);
''')

# Insert sample data
sample_data = [
    ('Germany', 'Europe', 357022, 83200000, 4200000000),
    ('Sweden', 'Europe', 450295, 10400000, 585000000),
    ('Norway', 'Europe', 385207, 5400000, 482000000),
    ('Denmark', 'Europe', 43094, 5800000, 395000000),
    ('France', 'Europe', 643801, 67500000, 2930000000),
    ('India', 'Asia', 3287263, 1400000000, 3170000000)
]

conn.executemany("INSERT INTO world VALUES (?, ?, ?, ?, ?)", sample_data)
conn.commit()

print("✅ World database loaded! Ready for SQL queries.")
```

## Day 1: SELECT Basics & Filtering

```python
query_1 = """
SELECT name, population 
FROM world 
WHERE name = 'Germany';
"""

# Display the output table
pd.read_sql_query(query_1, conn)
```

```python
query_2 = """
SELECT name, population 
FROM world 
WHERE name IN ('Sweden', 'Norway', 'Denmark');
"""

pd.read_sql_query(query_2, conn)
```

```python
import pandas as pd
import duckdb

data = [['0', 'Y', 'N'], ['1', 'Y', 'Y'], ['2', 'N', 'Y'], ['3', 'Y', 'Y'], ['4', 'N', 'N']]
products = pd.DataFrame(data, columns=['product_id', 'low_fats', 'recyclable']).astype({'product_id':'int64', 'low_fats':'category', 'recyclable':'category'})

sql_query = """
SELECT product_id 
FROM products 
WHERE low_fats = 'Y' AND recyclable = 'Y';
"""

duckdb.query(sql_query).df()
```
