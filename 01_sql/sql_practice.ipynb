### 🧩 LeetCode 1148: Article Views I
- **Difficulty:** Easy | **Platform:** LeetCode Top SQL 50
- **Concepts:** `SELECT DISTINCT`, Column Aliasing (`AS`), `WHERE`, `ORDER BY`
- **Goal:** Find authors who viewed at least one of their own articles (`author_id = viewer_id`), returning unique IDs sorted in ascending order.

```python
import pandas as pd
import duckdb

# 1. LeetCode Views Data Definition
data = [
    [1, 3, 5, '2019-08-01'],
    [1, 3, 6, '2019-08-02'],
    [2, 7, 7, '2019-08-01'],
    [2, 7, 6, '2019-08-02'],
    [4, 7, 1, '2019-07-22'],
    [3, 4, 4, '2019-07-21'],
    [3, 4, 4, '2019-07-21']
]
views = pd.DataFrame(data, columns=['article_id', 'author_id', 'viewer_id', 'view_date'])

# 2. Your SQL Query
sql_query = """
SELECT DISTINCT author_id AS id 
FROM views
WHERE author_id = viewer_id
ORDER BY id;
"""

# 3. Execute and display the formatted output table
duckdb.query(sql_query).df()
```

### 📊 Practical SQL Practice: Single-Column GROUP BY & Aggregations
- **Dataset:** `sales_orders` (E-Commerce transaction records)
- **Concepts:** `GROUP BY`, `COUNT(*)`, `SUM()`, `MIN()`, `MAX()`, Column Aliasing
- **Business Tasks:**
  1. **Category Price Extremes:** Find highest & lowest order amount per category.
  2. **Customer Spending:** Find total orders and revenue per customer.

```python
import duckdb

# 1. Setup sales_orders table
duckdb.sql('''
CREATE OR REPLACE TABLE sales_orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    category VARCHAR,
    order_amount DOUBLE,
    payment_method VARCHAR,
    city VARCHAR
);

INSERT INTO sales_orders VALUES
    (1, 201, 'Electronics', 1200.0, 'Credit Card', 'New York'),
    (2, 202, 'Clothing',     80.0,  'UPI',         'Chicago'),
    (3, 203, 'Electronics',  450.0, 'UPI',         'New York'),
    (4, 201, 'Electronics',  800.0, 'Credit Card', 'San Francisco'),
    (5, 204, 'Groceries',    150.0, 'Cash',        'Chicago'),
    (6, 205, 'Clothing',    200.0,  'Credit Card', 'New York'),
    (7, 206, 'Groceries',     90.0, 'UPI',         'San Francisco'),
    (8, 207, 'Electronics', 1500.0, 'Credit Card', 'Chicago'),
    (9, 208, 'Clothing',    120.0,  'UPI',         'New York'),
    (10, 209,'Groceries',    220.0, 'Cash',        'New York');
''')

# Task 1: Category Price Extremes
print('--- Task 1: Category Price Extremes (MAX & MIN) ---')
q1 = '''
SELECT 
    category, 
    MAX(order_amount) AS highest_order_amount, 
    MIN(order_amount) AS lowest_order_amount 
FROM sales_orders 
GROUP BY category;
'''
print(duckdb.sql(q1).df().to_string(index=False))

# Task 2: Customer Spending Habits
print('\n--- Task 2: Customer Spending Habits (COUNT & SUM) ---')
q2 = '''
SELECT 
    customer_id, 
    COUNT(*) AS no_of_orders, 
    SUM(order_amount) AS total_money_spent 
FROM sales_orders 
GROUP BY customer_id; 
'''
print(duckdb.sql(q2).df().to_string(index=False))
```

### 📈 SQL Practice: 06 ORDER BY & 07 HAVING
- **Datasets:** `customers` and `orders`
- **Concepts:** Sorting with `ASC`/`DESC`, Aggregation filtering with `HAVING` vs `WHERE`
- **Queries Solved:**
  1. **Alphabetical Sorting (ORDER BY):** Sort `customers` by name in Ascending & Descending order.
  2. **VIP Spending Filter (HAVING):** Group orders by `customer_id` and filter for total spending > 500.

```python
import duckdb

# 1. Setup customers and orders tables
duckdb.sql('''
CREATE OR REPLACE TABLE customers (
    customer_id INT PRIMARY KEY,
    customer_name VARCHAR,
    city VARCHAR
);

INSERT INTO customers VALUES 
    (201, 'Alice',   'New York'),
    (202, 'Bob',     'Chicago'),
    (203, 'Charlie', 'New York'),
    (204, 'David',   'San Francisco'),
    (205, 'Emma',    'Miami');

CREATE OR REPLACE TABLE orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    amount DOUBLE
);

INSERT INTO orders VALUES 
    (1, 201, 1200.00),
    (2, 201,  800.00),
    (3, 202,  150.00),
    (4, 203,  450.00),
    (5, 999,  300.00);
''')

# Q1: ORDER BY customer_name ASC & DESC
print('--- Q1: Customers Sorted Alphabetically (ASC) ---')
q1_asc = '''
SELECT customer_name 
FROM customers 
ORDER BY customer_name ASC;
'''
print(duckdb.sql(q1_asc).df().to_string(index=False))

print('\n--- Q1: Customers Sorted Alphabetically (DESC) ---')
q1_desc = '''
SELECT customer_name 
FROM customers 
ORDER BY customer_name DESC;
'''
print(duckdb.sql(q1_desc).df().to_string(index=False))

# Q2: GROUP BY + HAVING SUM(amount) > 500
print('\n--- Q2: Customers with Total Spending > 500 (HAVING) ---')
q2 = '''
SELECT customer_id, SUM(amount) AS total_amount 
FROM orders 
GROUP BY customer_id 
HAVING SUM(amount) > 500;
'''
print(duckdb.sql(q2).df().to_string(index=False))
```
